## Ejercicio 1 · Cargar el catálogo

El catálogo de productos ya está definido en la celda de identificación como la variable `CATALOGO` (lista de dicts). Implementa `ids_por_categoria(catalogo)` que devuelva un dict `{categoria: set_de_ids}`.

**Ejemplo:**
```
ids_por_categoria(CATALOGO)['tecnologia']
# -> {'P001', 'P002', 'P003', 'P008'}
```

Esta es la estructura de **índice por categoría** que permite recuperar todos los productos de una categoría en O(1).

````python
CATALOGO = [
    {'id': 'P001', 'nombre': 'laptop', 'categoria': 'tecnologia', 'precio': 2_500_000, 'stock': 15},
    {'id': 'P002', 'nombre': 'teclado', 'categoria': 'tecnologia', 'precio': 120_000, 'stock': 45},
    {'id': 'P003', 'nombre': 'monitor', 'categoria': 'tecnologia', 'precio': 800_000, 'stock': 8},
    {'id': 'P004', 'nombre': 'silla', 'categoria': 'hogar', 'precio': 450_000, 'stock': 20},
    {'id': 'P005', 'nombre': 'escritorio','categoria': 'hogar', 'precio': 900_000, 'stock': 5},
    {'id': 'P006', 'nombre': 'arroz', 'categoria': 'alimentos', 'precio': 5_000, 'stock': 200},
    {'id': 'P007', 'nombre': 'aceite', 'categoria': 'alimentos', 'precio': 18_000, 'stock': 150},
    {'id': 'P008', 'nombre': 'mouse', 'categoria': 'tecnologia', 'precio': 80_000, 'stock': 60},
    {'id': 'P009', 'nombre': 'lampara', 'categoria': 'hogar', 'precio': 75_000, 'stock': 30},
    {'id': 'P010', 'nombre': 'cafe', 'categoria': 'alimentos', 'precio': 25_000, 'stock': 80},
]
````

#PSEUDOCODIGO

      resultado ES IGUAL A diccionario vacio

      PARA CADA producto EN catalogo HACER
        categoria ES IGUAK A produccto["categoria"]
        codigo ES IGUAL A producto["id"]

        SI categoria ESTA EN resultado ENTONCES
          AGREGAR codigo A resultado[categoria]
        SINO
          resultado[categoria] ES IGUAL A {codigo}
        FIN SI
      FIN PARA

      RETORNAR resultado

````python
def ids_por_categoria(catalogo):
    resultado = {}

    for producto in catalogo:
      categoria = producto["categoria"]
      codigo = producto["id"]

      if categoria in resultado:
        resultado[categoria].add(codigo)
      else:
        resultado[categoria] = {codigo}
    return resultado
    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 1 ===
assert ids_por_categoria(CATALOGO)['tecnologia'] == {'P001', 'P002', 'P003', 'P008'}
assert ids_por_categoria(CATALOGO)['hogar'] == {'P004', 'P005', 'P009'}
print("✅ Ejercicio 1: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 1 ===
assert ids_por_categoria(CATALOGO)['alimentos'] == {'P006', 'P007', 'P010'}
assert set(ids_por_categoria(CATALOGO).keys()) == {'tecnologia', 'hogar', 'alimentos'}
assert ids_por_categoria([]) == {}
print("✅ Ejercicio 1: tests adicionales superados.")
````

## Ejercicio 2 · Índice id → producto

Implementa `indice_por_id(catalogo)` que devuelva un dict `{id: producto}` donde `producto` es el dict completo del producto.

Este índice permite recuperar todos los datos de un producto por su id en O(1),
en vez de recorrer el catálogo completo cada vez.

**Ejemplo:**
`indice_por_id(CATALOGO)['P003']['nombre']` → `'monitor'`

#PSEUDOCODIGO


      resultado ES IGUAL A diccionario vacio
      PARA CADA producto EN catalogo HACER
        codigo ES IGUAL A producto["id"]
        resultado[codigo] ES IGUAL A producto
      FIN PARA

````python
def indice_por_id(catalogo):
    resultado = {}

    for producto in catalogo:
      codigo = producto["id"]
      resultado[codigo] = producto
    return resultado
    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 2 ===
assert indice_por_id(CATALOGO)['P003']['nombre'] == 'monitor'
assert indice_por_id(CATALOGO)['P006']['precio'] == 5_000
print("✅ Ejercicio 2: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
assert set(indice_por_id(CATALOGO).keys()) == {p['id'] for p in CATALOGO}
assert indice_por_id(CATALOGO)['P001']['stock'] == 15
assert indice_por_id([]) == {}
print("✅ Ejercicio 2: tests adicionales superados.")
````

## Ejercicio 3 · Buscar por categoría

Implementa `buscar_categoria(catalogo, categoria)` que devuelva una **lista de dicts** con todos los productos de esa categoría.

Usa las funciones `ids_por_categoria` e `indice_por_id` que ya implementaste.

El resultado debe estar **ordenado por precio ascendente**.

Si la categoría no existe, devuelve lista vacía.

#PSEUDOCODIGO


      indice_categoria ES IGUAL A ids_por_categoria(catalogo)
      indice_producto ES IGUal  indice_por_id(catalogo)

      SI categoria NO ESTA EN indice categoria ENTONCES
        RETORNAR lista vacia
      FIN SI

      producto ES IGUAL A lista vacia

      PARA CADA codigo EN indice_categoria[categoria] HACER
        AGREGAR indice-producto[codigo] A productos
      FIN PARA

      ORDENAR oroductos POR precio DE MENOR A MAYOR

      RETORNAR productos

````python
def buscar_categoria(catalogo, categoria):
    indice_categoria = ids_por_categoria(catalogo)
    indice_producto = indice_por_id(catalogo)

    if categoria not in indice_categoria:
        return []

    productos = []

    for codigo in indice_categoria[categoria]:
        productos.append(indice_producto[codigo])

    productos.sort(key=lambda producto: producto["precio"])

    return productos
    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 3 ===
_alim = buscar_categoria(CATALOGO, 'alimentos')
assert len(_alim) == 3
assert _alim[0]['nombre'] == 'arroz'
assert _alim[-1]['nombre'] == 'cafe'
print("✅ Ejercicio 3: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
_tec = buscar_categoria(CATALOGO, 'tecnologia')
assert [p['id'] for p in _tec] == ['P008', 'P002', 'P003', 'P001']
assert buscar_categoria(CATALOGO, 'deportes') == []
print("✅ Ejercicio 3: tests adicionales superados.")
````

## Ejercicio 4 · Buscar por rango de precio

Implementa `buscar_rango_precio(catalogo, precio_min, precio_max)` que devuelva una **lista de dicts** con los productos cuyo precio esté entre `precio_min` y `precio_max` (inclusive).

El resultado debe estar ordenado por precio ascendente.

#PSEUDOCODIGO


      resultado ES IGUAL A lista vacia

      PARA CADA producto EN catalogo HACER

        PARA CADA producto EN catalogo HACER
          SI precio_min <= producto["precio"] <= precio_max ENTONCES
            AGREGAR producto a resultado
          FIN SI
        FIN PARA

        ORDENAR resultado POR precio de menor a mayor

        MOSTRAR resultado
      


````python
def buscar_rango_precio(catalogo, precio_min, precio_max):
    resultado = []

    for producto in catalogo:

        if precio_min <= producto["precio"] <= precio_max:
            resultado.append(producto)

    resultado.sort(key=lambda producto: producto["precio"])

    return resultado
    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 4 ===
_rango = buscar_rango_precio(CATALOGO, 50_000, 200_000)
assert len(_rango) == 3
assert _rango[0]['nombre'] == 'lampara'
print("✅ Ejercicio 4: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
assert buscar_rango_precio(CATALOGO, 0, 1_000) == []
_exacto = buscar_rango_precio(CATALOGO, 120_000, 120_000)
assert len(_exacto) == 1 and _exacto[0]['id'] == 'P002'
assert len(buscar_rango_precio(CATALOGO, 0, 9_999_999)) == len(CATALOGO)
print("✅ Ejercicio 4: tests adicionales superados.")
````

## Ejercicio 5 · Búsqueda multi-filtro

Implementa `buscar(catalogo, categoria=None, precio_min=None, precio_max=None, en_stock=False)` que aplique **todos** los filtros que no sean `None`:

- `categoria`: si se especifica, solo productos de esa categoría.
- `precio_min`: si se especifica, precio >= precio_min.
- `precio_max`: si se especifica, precio <= precio_max.
- `en_stock=True`: solo productos con stock > 0.

Devuelve lista de dicts ordenada por precio ascendente.

Usa las funciones anteriores como bloques de construcción o filtra directamente.

````python
def buscar(catalogo, categoria=None, precio_min=None, precio_max=None, en_stock=False):
    resultado = []

    for producto in catalogo:

        cumple = True

        if categoria is not None:
            if producto["categoria"] != categoria:
                cumple = False

        if precio_min is not None:
            if producto["precio"] < precio_min:
                cumple = False

        if precio_max is not None:
            if producto["precio"] > precio_max:
                cumple = False

        if en_stock:
            if producto["stock"] <= 0:
                cumple = False

        if cumple:
            resultado.append(producto)

    resultado.sort(key=lambda producto: producto["precio"])

    return resultado
    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 5 ===
_r1 = buscar(CATALOGO, categoria='hogar')
assert len(_r1) == 3
_r2 = buscar(CATALOGO, precio_min=100_000, precio_max=500_000)
assert all(100_000 <= p['precio'] <= 500_000 for p in _r2)
print("✅ Ejercicio 5: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 5 ===
_r3 = buscar(CATALOGO, categoria='tecnologia', precio_max=200_000)
assert set(p['id'] for p in _r3) == {'P002', 'P008'}
_r4 = buscar(CATALOGO, en_stock=True)
assert all(p['stock'] > 0 for p in _r4)
assert buscar(CATALOGO, categoria='deportes') == []
print("✅ Ejercicio 5: tests adicionales superados.")
````

## Ejercicio 6 · Resumen del catálogo

Implementa `resumen_catalogo(catalogo)` que devuelva un dict con el
resumen estadístico del catálogo, **reutilizando las funciones anteriores**:

```python
{
  'total_productos': int,
  'categorias': set,                # conjunto de categorías únicas
  'precio_min': int,                # precio del producto más barato
  'precio_max': int,                # precio del producto más caro
  'producto_mas_barato': str,       # nombre del producto de menor precio
  'productos_en_stock': int,        # cantidad con stock > 0
}
```

Si el catálogo está vacío, devuelve el dict con 0, set vacío, None donde corresponda.

````python
def resumen_catalogo(catalogo):
    if len(catalogo) == 0:
        return {
            "total_productos": 0,
            "categorias": set(),
            "precio_min": None,
            "precio_max": None,
            "producto_mas_barato": None,
            "productos_en_stock": 0
        }

    total_productos = len(catalogo)

    categorias = set()

    precios = []

    productos_en_stock = 0

    producto_mas_barato = None
    precio_min = None


    for producto in catalogo:

        categorias.add(producto["categoria"])

        precios.append(producto["precio"])


        if producto["stock"] > 0:
            productos_en_stock += 1


        if precio_min is None or producto["precio"] < precio_min:
            precio_min = producto["precio"]
            producto_mas_barato = producto["nombre"]


    precio_max = max(precios)


    return {
        "total_productos": total_productos,
        "categorias": categorias,
        "precio_min": precio_min,
        "precio_max": precio_max,
        "producto_mas_barato": producto_mas_barato,
        "productos_en_stock": productos_en_stock
    }

    raise NotImplementedError("Implementa esta función")
````

````python
# === Tests visibles · Ejercicio 6 ===
_res = resumen_catalogo(CATALOGO)
assert _res['total_productos'] == 10
assert _res['categorias'] == {'tecnologia', 'hogar', 'alimentos'}
assert _res['producto_mas_barato'] == 'arroz'
print("✅ Ejercicio 6: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
assert _res['precio_min'] == 5_000
assert _res['precio_max'] == 2_500_000
assert _res['productos_en_stock'] == 10
_res_vacio = resumen_catalogo([])
assert _res_vacio['total_productos'] == 0
assert _res_vacio['categorias'] == set()
print("✅ Ejercicio 6: tests adicionales superados.")
````
