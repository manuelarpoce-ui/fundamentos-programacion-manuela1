## Ejercicio 1 · Máximo Común Divisor (algoritmo de Euclides)

Implementa `mcd(a, b)` que devuelva el Máximo Común Divisor de dos enteros positivos usando el **algoritmo de Euclides**: mientras `b != 0`, reemplaza `(a, b)` por `(b, a % b)`; cuando `b == 0`, devuelve `a`.

**Ejemplos:** `mcd(48, 18)` → `6`; `mcd(100, 75)` → `25`; `mcd(13, 7)` → `1`.

#PSEUDOCODIGO


    MIENTRAS b SEA DIFERENTE A 0 HACER
      temporal <- b
      b <- a MOD b
      a <- temporal
    FIN MIENTRAS

    RETORNAR a

```python
def mcd(a, b):
    while b != 0:
      a,b=b, a%b
    return a
    raise NotImplementedError("Implementa esta función")
```

## Ejercicio 2 · Bubble sort (ordenamiento propio)

Implementa `bubble_sort(lista)` que ordene una lista de números de **menor a mayor** usando el algoritmo de burbuja: recorre la lista repetidamente e intercambia elementos adyacentes que estén en el orden incorrecto. Devuelve una **nueva lista** (no modifica la original).

**Ejemplo:** `bubble_sort([5, 2, 4, 1, 3])` → `[1, 2, 3, 4, 5]`.

Pista: una pasada completa 'burbujea' el mayor al final. Repite `n-1` pasadas.

#PSEUDOCODIGO


    lista_nueva <- copiar(lista)
    total_datos <- longitud(lista_nueva)

    PARA recorrido <- 0 HASTA total_datos - 2 HACER
      PARA indice <- 0 HASTA total_datos -2 - recorrido HACER
        SI lista_nueva[indice] > lista_nueva[indice +1] ENTONCES
          valor_temporal <- ista_nueva[indice]
          lista_nueva[indice] <- lista-nueva[indice+1]
          lista_nueva[indice+1] <- valor_temporal
        FIN SI
      FIN PARA
    FIN PARA

    RETORNAR lista_nueva

```python
def bubble_sort(lista):
    lista_nueva = lista.copy()
    total_datos = len(lista_nueva)
    for recorrido in range(total_datos - 1):
      for indice in range(total_datos - 1 - recorrido):
        if lista_nueva[indice] > lista_nueva[indice + 1]:
          valor_temporal = lista_nueva[indice]
          lista_nueva[indice] = lista_nueva[indice +1]
          lista_nueva[indice+1]=valor_temporal
    return lista_nueva
    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 2 ===
assert bubble_sort([5, 2, 4, 1, 3]) == [1, 2, 3, 4, 5]
assert bubble_sort([1]) == [1]
assert bubble_sort([]) == []
print("✅ Ejercicio 2: tests visibles superados.")
```

## Ejercicio 3 · Contar ocurrencias de cada letra

Implementa `contar_letras(texto)` que reciba una cadena (solo letras minúsculas y espacios) y devuelva una **lista de tuplas** `(letra, conteo)` ordenada **alfabéticamente** por letra. Ignora los espacios.

**Ejemplo:** `contar_letras("abba")` → `[('a', 2), ('b', 2)]`.

Pista: construye dos listas paralelas (letras únicas + conteos), luego ordena.

#PSEUDOCODIGO

      lista_letras <- lista vacia
      lista_cantidades <- lista vacia

      PARA cada caracter EN texto HACER
        SI caracter ES DIFERENTE A vacio ENTONCES
          SI caracter ESTA EN lista_letras ENTONCES
            posicion_letra <- posicion de caracter en lista_letras
            lista_cantidades[posicion_letra] <- lista_cantidades[posicion_letra]+1
          SINO
            Agregar caracter a lista_letras
            Agregar 1 a lista_cantidades
          FIN SI
        FIN SI
      FIN PARA

      lista_resultado <- zip(lista_letras, lista_cantidades)
      ordenar lista_resultado
      RETORNAR lista_resultado


```python
def contar_letras(texto):
    lista_letras=[]   #guarda la letra
    lista_cantidades=[]    #Guarda la cantidad de veces aparece la letra

    for caracter in texto:
      if caracter !=" ":    #Ignorar los espacios
        if caracter in lista_letras:    #revisar si la letra existe
          posicion_letra=lista_letras.index(caracter) #index nos dice en que posicion esta la letra
          lista_cantidades[posicion_letra]+=1 #Aumenta en 1 la cantidad de veces que aparece la letra

        else:     #Cuando la letra aparece por primera vez en la lista
          lista_letras.append(caracter)  #Si la letra es nueva agreguela al final
          lista_cantidades.append(1)  #Como es la primera vez que la encuentra contador inicia en 1
    lista_letras.sort()    #organiza una lista alfabeticamente
    lista_resultado=[]

    for caracter in lista_letras:
      total_repeticiones = texto.count(caracter)
      lista_resultado.append((caracter, total_repeticiones))

    return lista_resultado

    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 3 ===
assert contar_letras('abba') == [('a', 2), ('b', 2)]
assert contar_letras('aaa') == [('a', 3)]
print("✅ Ejercicio 3: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
assert contar_letras('') == []
assert contar_letras('cba') == [('a', 1), ('b', 1), ('c', 1)]
assert contar_letras('hola mundo') == [('a', 1), ('d', 1), ('h', 1), ('l', 1), ('m', 1), ('n', 1), ('o', 2), ('u', 1)]
print("✅ Ejercicio 3: tests adicionales superados.")
```

## Ejercicio 4 · Números primos hasta n (criba manual)

Implementa `primos_hasta(n)` que devuelva una **lista** con todos los números primos desde 2 hasta `n` inclusive. Un número es primo si solo es divisible por 1 y por sí mismo.

**Ejemplo:** `primos_hasta(20)` → `[2, 3, 5, 7, 11, 13, 17, 19]`.

Pista: escribe una función auxiliar `es_primo(k)` y usa un bucle externo.

```python
def es_primo(k):
    if k < 2:
      return False
    for divisor in range(2,k):
      if k % divisor == 0:
        return False
    return True

def primos_hasta(n):
    lista_primos = []
    for numero in range(2,n+1):
      if es_primo(numero):
        lista_primos.append(numero)
    return lista_primos


    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 4 ===
assert primos_hasta(20) == [2, 3, 5, 7, 11, 13, 17, 19]
assert primos_hasta(2) == [2]
assert primos_hasta(1) == []
print("✅ Ejercicio 4: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
assert primos_hasta(10) == [2, 3, 5, 7]
assert 97 in primos_hasta(100)
assert 100 not in primos_hasta(100)
print("✅ Ejercicio 4: tests adicionales superados.")
```
