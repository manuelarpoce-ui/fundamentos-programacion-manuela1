## Ejercicio 1 · Clasificar una edad (decisión)

Implementa `clasificar(edad)` que devuelva:
- `"menor"` si `edad < 18`
- `"adulto"` si `18 <= edad < 65`
- `"adulto mayor"` si `edad >= 65`

**Casos de ejemplo:** `clasificar(10)` → `"menor"`, `clasificar(40)` → `"adulto"`, `clasificar(70)` → `"adulto mayor"`.

#PSEUDOCODIGO

      SI edad < 18 ENTONCES
        RETORNE "menor"
      SINO SI edad < 65 ENTONCES
        RETORNE "adulto"
      SINO
        RETORNE "adulto manor
      FIN SI
      FIN

```python
def clasificar(edad):
    if edad < 18:
        return "menor"
    elif edad  < 65:
        return "adulto"
    else:
        return "adulto mayor"
    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 1 ===
assert clasificar(17) == 'menor'
assert clasificar(18) == 'adulto'
assert clasificar(64) == 'adulto'
assert clasificar(65) == 'adulto mayor'
print("✅ Ejercicio 1: tests adicionales superados.")
```

## Ejercicio 2 · Suma de los pares hasta n (bucle + condición)

Implementa `suma_pares(n)` que devuelva la suma de todos los números **pares** desde 1 hasta `n` (inclusive).

**Casos de ejemplo:** `suma_pares(6)` → `12` (2+4+6); `suma_pares(1)` → `0`.

#PSEUDOCODIGO

    suma_pares
      contador_suma <- 0
      PARA i desde 1 hasta n HACER
        SI i es numero par ENTONCES
          contador_suma = contador_suma + 1
        FINSI
      FIN PARA
    FIN



```python
def suma_pares(n):
    contador_suma = 0

    for i in range(1,n+1):
        if i % 2 == 0:
            contador_suma = contador_suma + i
    return contador_suma

    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 2 ===
assert suma_pares(6) == 12
assert suma_pares(1) == 0
print("✅ Ejercicio 2: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
assert suma_pares(2) == 2
assert suma_pares(10) == 30
assert suma_pares(0) == 0
print("✅ Ejercicio 2: tests adicionales superados.")
```

## Ejercicio 3 · Factorial (acumulador multiplicativo)

Implementa `factorial(n)` que devuelva `n!` = 1·2·3·…·n. Por convención, `factorial(0)` → `1`.

**Casos de ejemplo:** `factorial(5)` → `120`; `factorial(0)` → `1`.

#PSEUDOCODIGO

    respuesta <- 1

      PARA i desde 1 HASTA n HACER
          respuesta ES IGUAL A respuesta por i
      FIN PARA
    FIN

```python
def factorial(n):
    respuesta = 1   #Comienza en 1 porque todo numero multiplicado por 0 es cero

    for i in range(1, n+1):
        respuesta = respuesta * i
    return respuesta
    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 3 ===
assert factorial(5) == 120
assert factorial(0) == 1
print("✅ Ejercicio 3: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
assert factorial(1) == 1
assert factorial(3) == 6
assert factorial(7) == 5040
print("✅ Ejercicio 3: tests adicionales superados.")
```

## Ejercicio 5 · Segundo número más grande (dos campeones)

Implementa `segundo_mayor(lista)` que devuelva el **segundo** valor más grande de una lista con al menos 2 elementos. Si el máximo se repite, el segundo mayor puede ser igual al mayor.

**Casos de ejemplo:** `segundo_mayor([3, 9, 7, 9])` → `9`; `segundo_mayor([1, 2])` → `1`.

#PSEUDOCODIGO

iniciar numero mayor con un valor muy pequeño
INICIAR segundo con un valor my pequeño

      PARA cada numero en la lista HACER
        SI el numero ES MAYOR QUE numero_mayor ENTONCES
          segundo asignele numero_mayor
          numero_mayor asignele numero
        SINO SI el numero ES MAYOR QUE segundo ENTONCES
          segundo asignele numero
        FIN SI
      FIN PARA
      FIN

```python
def segundo_mayor(lista):
    numero_mayor = float ("-inf")   #float: numero con decimales
    segundo= float ("-inf")  # -inf: todavia no tengo un numero mayor, asi que inicio desd el mas pequeño

    for numero in lista:
        if numero > numero_mayor:
            segundo = numero_mayor
            numero_mayor = numero
        elif numero > segundo:
            segundo = numero
    return segundo
    raise NotImplementedError("Implementa esta función")
```

## Ejercicio 6 · Invertir una lista (sin reverse ni [::-1])

Implementa `invertir(lista)` que devuelva una **nueva** lista con los elementos en orden inverso, **sin** usar `lista[::-1]` ni `.reverse()`.

**Casos de ejemplo:** `invertir([1, 2, 3])` → `[3, 2, 1]`; `invertir([])` → `[]`.

#PSEUDOCODIGO
nueva_lista <- vacia

    PARA CADA elemento EN lista HACER:
      INSERTAR elemento EN LA POSICION 0 DE nueva_lista
    FIN PARA
      RETORNAR nueva_lista
    FIN

```python
def invertir(lista):
    nueva_lista = []

    for elemento in lista:              #Toma cada elemento de la lista uno por uno
        nueva_lista.insert(0, elemento) # insert() permite agregar un elemento en una poicion especifica
    return nueva_lista
    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 6 ===
assert invertir([1, 2, 3]) == [3, 2, 1]
assert invertir([]) == []
print("✅ Ejercicio 6: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
assert invertir([7]) == [7]
assert invertir(['a', 'b']) == ['b', 'a']
assert invertir([1, 1, 2]) == [2, 1, 1]
print("✅ Ejercicio 6: tests adicionales superados.")
```

## Ejercicio 7 · ¿Es primo? (bucle + salida temprana)

Implementa `es_primo(n)` que devuelva `True` si `n` es primo y `False` si no. Recuerda: los primos son enteros `>= 2` divisibles solo por 1 y por sí mismos. `0` y `1` no son primos.

**Casos de ejemplo:** `es_primo(7)` → `True`; `es_primo(1)` → `False`; `es_primo(9)` → `False`.

#PSEUDOCODIGO

    SI n < 2 ENTONCES
      RETORNE false
    FIN SI
    PARA i <- 2 HASTA n-1 HACER
      SI n MOD i ES IGUAL A 0 ENTONCES
        RETORNAR falso
    FIN SI
    FIN PARA

```python
def es_primo(n):
    if n < 2:    #Un numero primo es un numero entero mayor que 1 que solo tiene dos divisores
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
    raise NotImplementedError("Implementa esta función")
```

```python
# === Tests visibles · Ejercicio 7 ===
assert es_primo(7) is True
assert es_primo(1) is False
assert es_primo(9) is False
print("✅ Ejercicio 7: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 7 ===
assert es_primo(2) is True
assert es_primo(0) is False
assert es_primo(97) is True
assert es_primo(100) is False
print("✅ Ejercicio 7: tests adicionales superados.")
```
